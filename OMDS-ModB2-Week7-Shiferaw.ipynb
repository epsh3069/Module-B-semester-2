## Using Preattentive Attributes to Direct Attention to the Last 3 Months

### Suggested Techniques

1. **Fading the pre-period** : make the first 9 months lighter/transparent
   so the last 3 months naturally stand out by contrast

2. **Vertical breaking line** : a dashed vertical line at the 3-month mark
   creates a clear visual boundary between the two periods

3. **Directional arrows** : arrows starting at the breaking line and pointing
   toward the last 3 months explicitly guide the eye to the region of interest

4. **Value labels** : annotate the value at the start and at the 3-month
   mark to anchor the reader with context before they examine the final period

5. **Color contrast** — keep the post-period in a vivid color (e.g. steelblue)
   while rendering the pre-period in gray or muted tones

### Why These Work
Each technique exploits a different preattentive attribute:
- **Opacity/intensity** — faded elements recede visually
- **Enclosure/boundary** — the breaking line separates regions
- **Motion/direction** — arrows guide the eye explicitly
- **Color** — vivid colors are processed before muted ones
- **Text/annotation** — labels create focal anchor points

```python
import matplotlib.pyplot as plt
def preattentive_plot(datearray, time_series, title):
    df = pd.DataFrame({'data': time_series}, index=datearray)
   
    split_idx = int(len(datearray) - 3)
    split_date = df.index[split_idx]
    pre  = df.iloc[:split_idx+1]
    post = df.iloc[split_idx:]   # overlaps by 1 so the line connects

    start_val = df['data'].iloc[0]
    split_val = df['data'].iloc[split_idx]

    fig, ax = plt.subplots(figsize=(12, 5), facecolor='white')

    ax.plot(pre.index,  pre['data'],  color='steelblue', linewidth=2.5)
    ax.plot(post.index, post['data'], color='red', linewidth=2.5)

    ax.axvline(split_date, color='gray', linestyle='--', linewidth=1.5)

    ax.text(df.index[0], start_val + 2, f'{start_val:.0f}%', fontsize=9)
    ax.text(split_date,  split_val + 2, f'{split_val:.0f}%', fontsize=9)

    ax.set_title(f'{title} Level — Last 3 Months')
    ax.set_xlabel('Year and Month')
    ax.set_ylabel(f'{title} level (% of eutrophic value)')
    plt.tight_layout()
    plt.show()



numdates = 12
np.random.seed(0)
time_series = 100 * (np.random.normal(size=numdates) / 5 + np.arange(numdates) / 16)
datearray = pd.date_range(start='2023/01/01', end='2024/01/01', freq='M')
preattentive_plot(datearray, time_series, 'Nitrate')

time_series = 100 * (1 - (np.random.normal(size = numdates) / 8 + np.arange(numdates) / 16))
preattentive_plot(datearray, time_series, 'Phosphate')


```

# 4. Storytelling With Data plot

Reproduce any graph of your choice in p. 99-119 of the Storytelling With Data book as best you can. (The first half of chapter four). You do not have to get the exact data values right, just the overall look and feel.

```python
months = ['January', 'February', 'March', 'April', 'May', 'June',
          'July', 'August', 'September', 'October', 'November', 'December']

df_fake = pd.DataFrame({
    'month': months,
    
    'received':     [160, 185, 245, 150, 180, 165, 130, 200, 160, 135, 150, 175],
    'processed': [162, 183, 235, 148, 182, 150, 122, 157, 125, 105, 123, 140]
})
fig, ax = plt.subplots(figsize=(8, 6), facecolor='white')

received_color = 'gray'
processed_color = 'steelblue'
linewidth = 2

ax.plot(df_fake.index, df_fake['received'], 
       label='Ticket Volume Received', color=received_color, linewidth=linewidth)
ax.plot(df_fake.index, df_fake['processed'], 
       label='Ticket Volume Processed', color=processed_color, linewidth=linewidth)

for i in range(7, 12):
    ax.plot(i, df_fake['received'].iloc[i], marker='o', color=received_color, markersize=6)
    ax.plot(i, df_fake['processed'].iloc[i], marker='o', color=processed_color, markersize=6)
    ax.text(i, df_fake['received'].iloc[i]  + 8, str(df_fake['received'].iloc[i]),  
            ha='center', fontsize=9, color=received_color)
    ax.text(i, df_fake['processed'].iloc[i] - 16, str(df_fake['processed'].iloc[i]), 
            ha='center', fontsize=9, color=processed_color)

ax.set_facecolor('white')
ax.grid(axis='y', color='lightgray', linewidth=0.8)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.legend(loc='lower center', bbox_to_anchor=(0.5, -0.25), ncol=3)

ax.set_xticks(range(len(df_fake)))
ax.set_xticklabels(df_fake['month'].str[:3], ha='right')


# series labels at end of lines
ax.text(11.1, df_fake['received'].iloc[-1],  'Received',  fontsize=11, color=received_color,    va='center')
ax.text(11.1, df_fake['processed'].iloc[-1], 'Processed', fontsize=11, color=processed_color, va='center', fontweight='bold')


ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f'{x:.2f}'))
ax.set_ylim(bottom=0, top=300)

plt.tight_layout()
plt.show()
```
